```python
import gc

class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")

obj1 = MyObject("obj1")
obj2 = MyObject("obj2")
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2  #Circular reference

#Doing garbage collection manually
gc.collect()
```

```python
# Generators for memory efficiency

def generate_numbers(n):
    for i in range(n):
        yield i

# Generator in use
for num in generate_numbers(10000):
    print(num)
    if num>9:
        break
```

```python
# Profiling memory ussage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[Top 10]")
    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
